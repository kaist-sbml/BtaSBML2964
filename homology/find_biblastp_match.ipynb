```python
#e-value cutoff 1e-30으로 설정
import pandas as pd
df_cow_query = pd.read_csv('cow_query.tsv', sep = '\t')
df_human_query= pd.read_csv('human_query.tsv', sep = '\t')
```

```python
cow_query_dict= dict()
human_query_dict= dict()
for index, row in df_cow_query.iterrows():
    if row['query'] not in cow_query_dict:
        cow_query_dict[row['query']]= [row['data'], row['e_value']]
    else:
        original_value= cow_query_dict[row['query']][1]
        if row['e_value'] < original_value:
            cow_query_dict[row['query']]= [row['data'],row['e_value']]
            
for index, row in df_human_query.iterrows():
    if row['query'] not in human_query_dict:
        human_query_dict[row['query']]= [row['data'], row['e_value']]
    else:
        original_value= human_query_dict[row['query']][1]
        if row['e_value'] < original_value:
            human_query_dict[row['query']]= [row['data'],row['e_value']]
```

```python
match_dict= dict()
for query, match_inform in cow_query_dict.items():
    human_gene_name= match_inform[0]
    if human_gene_name in human_query_dict: 
        cow_gene_name= human_query_dict[human_gene_name][0]
        if query == cow_gene_name:
            match_dict[human_gene_name]= cow_gene_name
```

```python
import pickle
with open('biblastp_match.pickle', 'wb') as f:
    pickle.dump(match_dict,f)
```
