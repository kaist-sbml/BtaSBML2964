```python
from Bio import SeqIO

genomes = SeqIO.parse("genomic.gbk", "genbank")

cds = dict()
for genome in genomes:
    for feature in genome.features:
        if feature.type == 'CDS':
            if 'gene' in feature.qualifiers and 'translation' in feature.qualifiers:
                cds[feature.qualifiers['gene'][0]]= feature.qualifiers['translation'][0]
        
f=open('cow_gene_seq.fasta','w')
for gene, seq in cds.items():
    f.write('>'+gene+'\n')
    f.write(seq+'\n')
f.close()
```
